```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("s254232.csv")

ratings = pd.concat([df["Rating 1"], df["Rating 2"]]).dropna()

min_rating = int(ratings.min())
max_rating = int(ratings.max())

bins = np.arange(min_rating - 0.5, max_rating + 1.5, 1)

plt.hist(ratings, bins=bins, edgecolor="black")
plt.xticks(range(min_rating, max_rating + 1))
plt.xlabel("Rating")
plt.ylabel("Number of ratings")
plt.title("Distribution of all ratings")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("average_ratings.csv")

# Get the average ratings
ratings = df["average_rating"].dropna()

# Since averages of 4 ratings can occur in steps of 0.25
min_rating = ratings.min()
max_rating = ratings.max()

bins = np.arange(min_rating - 0.125, max_rating + 0.375, 0.25)

plt.hist(ratings, bins=bins, edgecolor="black")

plt.xlabel("Average Rating")
plt.ylabel("Number of Images")
plt.title("Distribution of Average Ratings")

plt.show()
```

```python
import pandas as pd

df = pd.read_csv("average_ratings.csv")

# Min-max normalize average_rating to range 0-1
min_rating = df["average_rating"].min()
max_rating = df["average_rating"].max()

df["normalized_rating"] = (
    df["average_rating"] - min_rating
) / (
    max_rating - min_rating
)

# Keep image name and normalized rating
normalized_df = df[["Images", "normalized_rating"]]

normalized_df.to_csv("normalized_ratings.csv", index=False)

print("Saved normalized_ratings.csv")
```
