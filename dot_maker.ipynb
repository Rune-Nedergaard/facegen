```python
from PIL import Image, ImageDraw
import os

folder = "adaptors"
dot_radius = 1  # radius in pixels

for filename in os.listdir(folder):

    if filename.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".webp")):

        path = os.path.join(folder, filename)

        with Image.open(path) as img:

            img = img.convert("RGB")
            draw = ImageDraw.Draw(img)

            width, height = img.size

            center_x = width // 2
            center_y = height // 2

            # Draw black dot
            draw.ellipse(
                (
                    center_x - dot_radius,
                    center_y - dot_radius,
                    center_x + dot_radius,
                    center_y + dot_radius,
                ),
                fill="red"
            )

            img.save(path)

        print(f"Added fixation dot to: {filename}")

print("Done!")
```
