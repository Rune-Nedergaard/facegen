```python
import pandas as pd

# Load the two CSV files
df1 = pd.read_csv("s254232.csv")
df2 = pd.read_csv("member_2.csv")

# Find images that appear in BOTH files
common_images = set(df1["Image"]) & set(df2["Image"])

# Keep only those images
df1_clean = df1[df1["Image"].isin(common_images)]
df2_clean = df2[df2["Image"].isin(common_images)]

# Save cleaned CSV files
df1_clean.to_csv("ratings1_clean.csv", index=False)
df2_clean.to_csv("ratings2_clean.csv", index=False)

print(f"Images in file 1 originally: {len(df1)}")
print(f"Images in file 2 originally: {len(df2)}")
print(f"Images present in both: {len(common_images)}")
```

```python
import pandas as pd

# Load the two cleaned CSV files
df1 = pd.read_csv("ratings1_clean.csv")
df2 = pd.read_csv("ratings2_clean.csv")

# Rename columns
df1 = df1.rename(columns={
    "Image": "Images",
    "Rating 1": "rating1",
    "Rating 2": "rating2"
})

df2 = df2.rename(columns={
    "Image": "Images",
    "Rating 1": "rating3",
    "Rating 2": "rating4"
})

# Merge based on image name
merged_df = pd.merge(df1, df2, on="Images", how="inner")

# Calculate the mean rating for each image
merged_df["average_rating"] = merged_df[
    ["rating1", "rating2", "rating3", "rating4"]
].mean(axis=1)

# Keep only image name and average rating
merged_df = merged_df[["Images", "average_rating"]]

# Save result
merged_df.to_csv("average_ratings.csv", index=False)

print(f"Calculated average ratings for {len(merged_df)} images.")
```
